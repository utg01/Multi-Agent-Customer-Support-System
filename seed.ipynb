```python
#ORDERS
def create_fake_order(db,user,prod_objs):

    order=Order(user_id=user.user_id, total_amount=0)
    db.add(order)
    db.flush()

    ordered_products=random.sample(prod_objs, k=random.randint(1,3))
    total=0
    for p in ordered_products:
        quantity=random.randint(1,3)
        item=OrderItem(
            order_id=order.order_id,
            product_id=p.prod_id,
            quantity=quantity,
            price_at_purchase=p.price
        )
        db.add(item)
        total+=quantity*p.price
    order.total_amount=total
    return order

for i in range (150):
    random_user=random.choice(users)
    create_fake_order(db,random_user,prod_objs)

db.commit()
```
